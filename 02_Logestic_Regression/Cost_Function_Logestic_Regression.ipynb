```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
    
```

```python
def compute_cost_logistic(X,y,w,b):

    m=X.shape[0]

    cost=0
    for i in range(m):
        z_i=X[i]@w+b
        f_wb_i=sigmoid(z_i)
        cost+=-y[i]*np.log(f_wb_i)-(1-y[i])*np.log(1-f_wb_i)
    cost=cost/m


    return cost
    
        
```

```python
X_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])  #(m,n)
y_train = np.array([0, 0, 0, 1, 1, 1])                                           #(m,)
```

```python
def plot_data(X, y):
    # Separate the classes
    X_pos = X[y == 1]
    X_neg = X[y == 0]

    # Positive examples
    plt.scatter(
        X_pos[:, 0], X_pos[:, 1],
        marker='x',
        s=180,
        color='red',
        linewidths=4,
        label='y=1'
    )

    # Negative examples
    plt.scatter(
        X_neg[:, 0], X_neg[:, 1],
        marker='o',
        s=180,
        facecolors='none',
        edgecolors='dodgerblue',
        linewidths=3,
        label='y=0'
    )
```

```python
plot_data(X_train, y_train)
plt.xlabel(r'$X_0$', fontsize=16)
plt.ylabel(r'$X_1$', fontsize=16)

# Limits
plt.xlim(0,4)
plt.ylim(0,3.5)

# Legend
plt.legend()
plt.show()
```

```python
w_tmp = np.array([1,1])
b_tmp = -3
print(compute_cost_logistic(X_train, y_train, w_tmp, b_tmp))
```

```python
plot_data(X_train, y_train)
plt.xlabel(r'$X_0$', fontsize=16)
plt.ylabel(r'$X_1$', fontsize=16)

# Limits
plt.xlim(0,4)
plt.ylim(0,3.5)

# Legend
plt.legend()

#decision boundaries

x=np.linspace(0,4,100)

w=np.array([1,1])

#b=-3

b=-3
y_line=-(w[0]*x+b)/w[1]
plt.plot(x, y_line, lw=4, label='b=-3')

# b = -4
b = -4
y_line = -(w[0]*x + b)/w[1]
plt.plot(x, y_line, lw=4, color='orchid', label='b=-4')

# ----------------------------

plt.xlim(0,4)
plt.ylim(0,4)

plt.xlabel(r'$X_0$', fontsize=16)
plt.ylabel(r'$X_1$', fontsize=16)
plt.title("Decision Boundary", fontsize=20)

plt.legend()
plt.grid(False)

plt.show()
```

```python
w_array1 = np.array([1,1])
b_1 = -3
w_array2 = np.array([1,1])
b_2 = -4

print("Cost for b = -3 : ", compute_cost_logistic(X_train, y_train, w_array1, b_1))
print("Cost for b = -4 : ", compute_cost_logistic(X_train, y_train, w_array2, b_2))
```
