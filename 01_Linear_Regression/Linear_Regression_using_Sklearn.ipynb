```python
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import numpy as np
```

```python
X = np.array([
    [2, 1800, 3, 8],
    [5, 1500, 3, 5],
    [12, 1200, 2, 3],
    [20, 1000, 2, 1],
    [8, 2000, 4, 10],
    [15, 1600, 3, 6],
    [25, 900, 2, 2],
    [10, 1400, 3, 7],
    [30, 1100, 2, 4],
    [3, 2200, 4, 12],
    [18, 1700, 3, 8],
    [22, 1300, 2, 5],
    [6, 1900, 4, 9],
    [14, 1550, 3, 4],
    [28, 950, 1, 2],
    [9, 1750, 3, 11],
    [17, 1450, 3, 6],
    [24, 1250, 2, 3],
    [7, 2100, 4, 13],
    [35, 850, 1, 1]
])

y = np.array([
    365, 310, 220, 170, 420,
    300, 145, 285, 180, 470,
    320, 230, 405, 295, 130,
    350, 275, 205, 450, 110
])

X_features=["Age", "Size", "Bedrooms", "Floor"]
```

```python
scaler=StandardScaler()
X_Norm=scaler.fit_transform(X)
X_Norm
```

```python
fig,ax=plt.subplots(1,4,figsize=(12,4),sharey=True)
for i in range(len(X[0])):
    ax[i].scatter(X[:,i],y)
    ax[i].set_xlabel(X_features[i])
ax[0].set_ylabel('price')
    
```

```python
model=SGDRegressor(max_iter=10000,eta0=0.01,learning_rate='constant')
model.fit(X_Norm,y)
print(model.coef_)
print(model.intercept_)
```

```python
yp=model.predict(X_Norm)
fig,ax=plt.subplots(1,4,figsize=(12,4),sharey=True)
for i in range(len(X[0])):
    ax[i].scatter(X[:,i],y,label='target')
    ax[i].set_xlabel(X_features[i])
    ax[i].scatter(X[:,i],yp,label='predicted')
    
ax[0].set_ylabel('price')
ax[0].legend()
plt.show()
```
